## 基本使用範例

以下是 MQTT 客戶端的基本使用方式：


````python
import paho.mqtt.client as mqtt
import time

# MQTT 伺服器設定（請根據您的實際設定修改）
BROKER = "localhost"  # 或您的 MQTT broker 位址，例如 "192.168.1.100"
PORT = 1883           # 預設 MQTT 埠號
TOPIC = "test/topic"  # 要訂閱/發布的主題

# 連線回呼函數
def on_connect(client, userdata, flags, rc):
    if rc == 0:
        print("連線成功！")
        client.subscribe(TOPIC)  # 訂閱主題
    else:
        print(f"連線失敗，錯誤代碼: {rc}")

# 接收訊息回呼函數
def on_message(client, userdata, msg):
    print(f"收到訊息 - 主題: {msg.topic}, 內容: {msg.payload.decode()}")

# 建立 MQTT 客戶端
client = mqtt.Client()
client.on_connect = on_connect
client.on_message = on_message

# 連線到 MQTT broker
try:
    client.connect(BROKER, PORT, 60)
    print(f"嘗試連線到 {BROKER}:{PORT}...")
except Exception as e:
    print(f"連線錯誤: {e}")

# 開始網路循環（非阻塞模式）
client.loop_start()

# 等待一下讓連線建立
time.sleep(1)

# 發布訊息
message = "Hello MQTT from Raspberry Pi!"
client.publish(TOPIC, message)
print(f"已發布訊息: {message}")

# 等待接收訊息
time.sleep(5)

# 停止網路循環
client.loop_stop()
client.disconnect()
print("已斷線")
````

## 其他可選的 MQTT 套件
